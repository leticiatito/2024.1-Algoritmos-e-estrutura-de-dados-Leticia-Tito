```python
#Verifica se a string `palavra` é um palíndromo, ignorando espaços e
#considerando letras maiúsculas e minúsculas como iguais.

def eh_palindromo(palavra):
  #pass
  palavra = palavra.replace(" ", "").lower()
  return palavra == palavra[::-1]

print(eh_palindromo("arara"))
print(eh_palindromo("subi no onibus"))
print(eh_palindromo("algoritmos"))
```
