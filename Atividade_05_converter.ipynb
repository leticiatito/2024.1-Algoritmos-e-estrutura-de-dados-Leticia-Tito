```python
#converte o conteudo do arquivo 'arquivo_entrada' para maiusculas e salva o resultado em 'arquivo_saida'

def converter_para_maiusculas(arquivo_entrada, arquivo_saida):
  #pass
  with open('/content/drive/MyDrive/Colab Notebooks/arquivo_entrada.txt', 'r') as arquivo_entrada:
    conteudo = arquivo_entrada.read()
    conteudo_maiusculas = conteudo.upper()
    with open('/content/drive/MyDrive/Colab Notebooks/arquivo_saida.txt', 'w') as arquivo_saida:
      arquivo_saida.write(conteudo_maiusculas)
      print('Arquivo convertido com sucesso')


arquivo_entrada = '/content/drive/MyDrive/Colab Notebooks/arquivo_entrada.txt'
arquivo_saida = '/content/drive/MyDrive/Colab Notebooks/arquivo_saida.txt'
converter_para_maiusculas('arquivo_entrada.txt', 'arquivo_saida.txt')

```
