```python
#conta a frequencia de cada caractere na string 's'
def contar_caracteres(s):
    contagem = {}
    for caractere in s:
        if caractere in contagem:
            contagem[caractere] += 1
        else:
            contagem[caractere] = 1
    return contagem

texto1 = "abcabc"
resultado1 = contar_caracteres(texto1)
print(resultado1)

texto2 = "a b c a b c"
resultado2 = contar_caracteres(texto2)
print(resultado2)

texto3 = "/"
resultado3 = contar_caracteres(texto3)
print(resultado3)
```
