```python
#conta a frequencia de cada caractere na string 's'
def contar_caracteres(s):
    contagem = {}
    for caractere in s:
        if caractere in contagem:
            contagem[caractere] += 1
        else:
            contagem[caractere] = 1
    return contagem

texto = "hello, world!"
resultado = contar_caracteres(texto)
print(resultado)
```
