```python
#conta quantas vezes a palavra "teste" aparece no arquivo "teste.txt"
def contar_palavra(arquivo, palavra):
    with open("/content/drive/MyDrive/Colab Notebooks/teste.txt", 'r') as arquivo:
        conteudo = arquivo.read()
        conteudo = conteudo.lower()
        palavras = conteudo.split()
        contador = 0
        for p in palavras:
            if p == palavra:
                contador += 1
    return contador

print(contar_palavra("/content/drive/MyDrive/Colab Notebooks/teste.txt", "teste"))
print(contar_palavra("/content/drive/MyDrive/Colab Notebooks/teste.txt", "algoritmos"))
```
